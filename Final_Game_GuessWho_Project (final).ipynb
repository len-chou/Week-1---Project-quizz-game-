```python
import random  # load Python's random module so we can pick something randomly later


# Dictionary containing 3 clues for each person
# each key is a name, each value is a list of 3 clues strings
clues = {
    "Gabby Williams": [
        "I am a WNBA All-Star",
        "I am a female athlete",
        "I am French and American",
    ],

    "Celine Dion": [
        "I am a female singer",
        "I am from Quebec",
        "One of my songs is featured in the movie Titanic"
    ],

    "David Guetta": [
        "I am a French DJ",
        "I mostly perform in Spain Ibiza",
        "I did a lot of featuring with other famous artists"
    ],

    "Antoine Dupont": [
        "I am a male French athlete",
        "I play a collective sport",
        "My sport is rugby"
    ]
}


# This function contains the entire game, from start to finish
def play_game():

    # Make a list with just the names (keys)
    # This list will shrink each time someone is guessed, since we remove them after each round
    famous_people = list(clues.keys())

    # This will add up points across every round
    total_points = 0

    print("Welcome to the GUESS WHO I AM !")
    print()
    print("Goal of the game: \n You will need to uncover the chosen celebrity, with clues that will be given to you for each rounds to help you. \nIn total, you have 3 clues per celebrities.")
    print("After each clue, you will have to find who the celebrity is. After the last clue if the answer is WRONG you will loose.")
    print()
    print("Good Luck and Let's start with the first clue!\n")
    print()
    print()

    # Outer loop: keep playing as long as there is at least one name left in the list
    while len(famous_people) > 0:

        # Pick one name randomly from whoever is left.
        secret_person = random.choice(famous_people)

        # Get the list of clues for that person
        person_clues = clues[secret_person]

        # This will count how many wrong guesses we made this round
        wrong_guesses = 0

        # Have we found the answer yet? Start with False
        found = False

        # Repeat this block 3 times (index will be 0, then 1, then 2)
        for index in range(3):

            print(person_clues[index])  # show the clue at the current index
            print()

            guess = input("Who do you think it is? ")  # ask the player to type a guess
            print()

            if guess.lower() == secret_person.lower():
                found = True  # mark that the player got it right
                break  # stop the loop immediately, no need to show more clues

            else:
                wrong_guesses = wrong_guesses + 1  # add 1 to the wrong-guess counter
                print()
                print("Wrong guess. Wrong answers:", wrong_guesses)
                print()

        # This runs after the inner loop ends, whether it stopped early (break) or ran out of clues
        if found == True:
            print()
            print("Correct! It was", secret_person)
            print()

            # Add this round's wrong guesses to the running total
            total_points = total_points + wrong_guesses

            # Remove this person from the list so they can't be picked again
            famous_people.remove(secret_person)

        else:
            print("Game over. The answer was", secret_person)
            print()

            # Add this round's wrong guesses to the running total
            total_points = total_points + wrong_guesses

            # Stop the entire game immediately - do NOT move to the next celebrity
            break

    # This runs once the outer loop is done, either everyone was guessed or the game ended early
    print()
    print("Total wrong guesses:", total_points)
    print()


# Call the function to actually start the game play_game()
```

```python
play_game()
```
